```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
np.__version__ , matplotlib.__version__
```

```python
def game(qs_mean, k : int , epoch : int, eps : float = 0.1):
  Q_est = np.zeros(k)
  a_times = np.zeros(k)
  for t in range(epoch):
    if np.random.rand() < eps:
      a = np.argmax(Q_est)
    else:
      a = np.random.randint(k)
    r = qs_mean[a] + np.random.randn()
    Q_est[a] = (Q_est[a] * a_times[a] + r) / (a_times[a] + 1)
    a_times[a] += 1
  return Q_est

def visualize(Q_est : np.ndarray, qs_ref : np.ndarray):
  plt.plot(Q_est)
  plt.plot(qs_ref)
  plt.legend(['Q_est', 'qs_ref'])
  plt.show()
```

```python
k = 100
qs_mean = np.random.randn(k)/10 + 1
```

```python
Q_est = game(qs_mean, k, 100000, 0)
visualize(Q_est, qs_mean)
```
